```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

# Define the vertices of the probability simplex
simplex_vertices = np.array([
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1]
])

# Define some example points in the generalized affine space
example_points = np.array([
    [0.5, 0.5, 0],    # Point on the edge
    [0.7, 0.3, 0],    # Point inside the simplex
    [1.2, -0.2, 0],   # Point outside the simplex
    [-0.2, 0.4, 0.8]  # Point with a negative coordinate
])

# Function to plot the probability simplex
def plot_simplex(ax):
    # Create a Poly3DCollection for the simplex
    poly = Poly3DCollection([simplex_vertices], alpha=0.5, facecolor='blue')
    ax.add_collection3d(poly)

    # Plot the vertices of the simplex
    ax.scatter(simplex_vertices[:, 0], simplex_vertices[:, 1], simplex_vertices[:, 2], color='blue', s=100)
    
    # Plot the edges of the simplex
    for i in range(len(simplex_vertices)):
        for j in range(i+1, len(simplex_vertices)):
            ax.plot([simplex_vertices[i, 0], simplex_vertices[j, 0]],
                    [simplex_vertices[i, 1], simplex_vertices[j, 1]],
                    [simplex_vertices[i, 2], simplex_vertices[j, 2]], color='blue')

# Plot the generalized affine space
def plot_example_points(ax, points):
    # Plot the example points
    ax.scatter(points[:, 0], points[:, 1], points[:, 2], color='red', s=100)
    for point in points:
        ax.text(point[0], point[1], point[2], f'({point[0]:.1f}, {point[1]:.1f}, {point[2]:.1f})', color='red')

# Create a 3D plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot the probability simplex
plot_simplex(ax)

# Plot the example points
plot_example_points(ax, example_points)

# Set labels
ax.set_xlabel('$v_1$')
ax.set_ylabel('$v_2$')
ax.set_zlabel('$v_3$')

# Set limits
ax.set_xlim(-0.5, 1.5)
ax.set_ylim(-0.5, 1.5)
ax.set_zlim(-0.5, 1.5)

# Set title
#ax.set_title('Probability Simplex and Generalized Affine Space Example Points')

# Add caption
#fig.text(0.5, 0.01, 'Figure 1: The blue triangle represents the standard probability simplex. Red points represent vectors in the generalized affine space, including points with negative values or values exceeding 1.', ha='center', fontsize=10)

plt.show()
```
